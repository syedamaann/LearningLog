```python
from os import read
# Write a generator-based pipeline to:

# Read a file line by line.
# Filter lines containing a specific keyword.
# Extract and yield the first word from each filtered line.

def read_file(filename):
  with open(filename) as file:
    for line in file:
      yield line

def filter_lines(lines, keyword):
  for line in lines:
    if keyword in line:
      yield line

def extract_first_word(lines):
  for line in lines:
    yield line.split()[0]


# Use yield and generator chaining.

for word in extract_first_word(filter_lines(read_file("logs.text"), "ERROR")):
  print(word)
```

# Decorators and Higher-Order Functions

```python
def greet(name):
  return f"Hello, {name}!"

def execute(func, name):
  return func(name)

execute(greet, "alice")
```

```python
def logger(func):
  def wrapper(*args, **kwargs):
    print(f"calling {func.__name__} with {args}")
    print(func(*args, **kwargs))
    print("line after func")
  return wrapper

@logger
def add(a,b):
  return a+b

(add(3,5))
```

```python
class MultiplyBy:
  def __init__(self, factor):
    self.factor = factor

  def __call__(self, func):
    def wrapper(*args, **kwargs):
      print(f"calling {func.__name__} with {args} and factor {self.factor}")
      print(func(*args,**kwargs) * self.factor)
    return wrapper

@MultiplyBy(2)
def add(a,b):
  return a+b

add(3,5)
```

```python
import time

def timer(func):
  def wrapper(*args, **kwargs):
    start = time.time()
    print(f"calling {func.__name__} with {args}")
    print(func(*args, *kwargs))
    end = time.time()
    print(f"time for {func.__name__}: {end - start}")
  return wrapper

@timer
def add(a,b):
  time.sleep(3)
  return a+b

add(3,5)
```

```python
def log_arguments(func):
  def wrapper(*args, **kwargs):
    print(f"calling {func.__name__} with args: {args}, kwargs: {kwargs}")
    result = func(*args, **kwargs)
    print(f"{func.__name__} func returned {result}")
    return result
  return wrapper

@log_arguments
def add(a,b):
  return a+b

add(10,9)
```

```python
user1 = {"name": "alice", "age": 23, "authenticated": True}
user2 = {"name": "bob", "age": 35, "authenticated": False}

def require_authentication(func):
  def wrapper(user, *args, **kwargs):
    if not user.get("authenticated"):
      return f"user not authenticated to view this page"
    return func(user, *args, **kwargs)
  return wrapper

@require_authentication
def view_dashbboard(user):
  print(f"welcome to the wonderful dashboard, {user['name']}!")

print(view_dashbboard(user2))
view_dashbboard(user1)
```

```python
def cache(func):
    memo = {}
    def wrapper(*args):
        if args in memo:
            print(f"Fetching cached result for {args}")
            return memo[args]
        print(f"Computing result for {args}")
        result = func(*args)
        memo[args] = result
        end = time.time()
        return result
    return wrapper

@cache
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(10))
# Output:
# Computing result for (10)
# ...
# Computing result for (2)
# Computing result for (1)
# Fetching cached result for (2)
# 55
```

```python
def cache(func):
  memo = {}
  def wrapper(*args, **kwargs):
    if args in memo:
      print(f"cache hit for {args}")
      return memo[args]
    print(f"computing for {args}")
    result = func(*args, **kwargs)
    memo[args] = result
    return result
  return wrapper


@cache
def fib(n):
  if n <= 1:
    return n
  return fib(n-1)+fib(n-2)

fib(10)
```

```python
def counter(func):
  func.call_count = 0
  def wrapper(*args, **kwargs):
    func.call_count += 1
    print(f"{func.__name__ } is called {func.call_count} times")
    print(func(*args,**kwargs))
  return wrapper

@counter
def add(a,b):
  return a+b

add(2,3)
add(4,5)
add(67,78)
add(67,78)
```

```python
def upper_case(func):
  def wrapper(*args, **kwargs):
    result = func(*args, **kwargs)
    return result.upper()
  return wrapper

@upper_case
def greet(name):
  return f"hello {name}"

print(greet("amaan"))
```

```python
import random

def retry(times):
  def decorator(func):
    def wrapper(*args, **kwargs):
      for _ in range(times):
        try:
          result = func(*args, **kwargs)
          return result
        except Exception:
          print(f"retrying {func.__name__}")
    return wrapper
  return decorator

@retry(times=3)
def unstable_func():
  if random.random() < 0.7:
    raise Exception
  return "success!"

unstable_func()
```

```python
def repeat(times):
  def decorator(func):
    def wrapper(*args, **kwargs):
      for _ in range(times):
        func(*args, **kwargs)
    return wrapper
  return decorator

@repeat(3)
def greeting(name):
  print(f"hello {name}")


greeting("amaan")
```

```python
class BankAccount:
  def __init__(self, owner, balance):
    self.owner = owner
    self.balance = balance

    @property
    def balance(self):
      return self.balance

    @balance.setter
    def balance(self, value):
      if value < 0:
        raise ValueError("balance cannot be negative")
      self._balance = value

account = BankAccount("amaan", 10000)
print(account.balance)
account.balance = 200000
account.balance
```

```python
class Rectangle:
  def __init__(self, length, width):
    self.length = length
    self.width = width

  def area(self):
    return self.length * self.width

  def perimeter(self):
    return 2*(self.length + self.width)

  def __str__(self) -> str:
    return f"area {self.area()}, perimeter {self.perimeter()}"

rect = Rectangle(2,4)

print(rect)
```

```python
class BankAccount:
  def __init__(self, owner, balance):
    self.owner = owner
    self.balance = balance

  def deposit(self, amount):
    self.balance += amount
    return f"deposited {amount} and your new balance is {self.balance}"

  def withdraw(self, amount):
    if self.balance < amount:
      raise ValueError("insufficient balance")
    self.balance -= amount
    return f"withdrew {amount} and your new balance is {self.balance}"

  def get_statement(self):
    return f"you are {self.owner} and your account balance is {self.balance} "

account = BankAccount("amaan", 100000)
account.deposit(200)
account.withdraw(500)
account.get_statement()
```

```python
import heapq
class TaskManager:
  def __init__(self):
    self.tasks = []

  def add_task(self, description, priority):
    heapq.heappush(self.tasks, (priority, description))

  def remove_task(self, description):
    self.tasks = [task for task in self.tasks if task[1] != description]

  def get_tasks_by_priority(self, priority):
    return [task for task in self.tasks if task[0] == priority]

  def __str__(self) -> str:
    return f"tasks: {self.tasks}"

taskmanager = TaskManager()
taskmanager.add_task('task1', 'high')
taskmanager.add_task('task2', 'low')
taskmanager.add_task('task3', 'medium')
taskmanager.add_task('task4', 'low')
taskmanager.add_task('task5', 'high')
taskmanager.add_task('task6', 'medium')
taskmanager.remove_task('task2')
print(taskmanager.get_tasks_by_priority('high'))
print(taskmanager)
```

```python
class Vehicle:
  def __init__(self, make, model, year):
    self.make = make
    self.model = model
    self.year = year

class Car(Vehicle):
  def __init__(self, make, model, year):
    super().__init__(make, model, year)

  def drive(self, km):
    self.km = km
    return f"todat i drove my {self.make} of {self.model} from {self.year} for {self.km}km"

class Motorcycle(Vehicle):
  def __init__(self, make, model, year):
    super().__init__(make, model, year)

  def ride(self):
    return f"riding a {self.make} {self.model} {self.year}"

class Truck(Vehicle):
  def __init__(self, make, model, year):
    super().__init__(make, model, year)

  def load_capacity(self, capacity):
    return f"the load capacity of the {self.make} {self.model} {self.year} is {capacity}"

truck = Truck("Toyota", "Camry", 2023)
print(truck.load_capacity(10000))

motorcycle = Motorcycle("Honda", "CBR", 2023)
print(motorcycle.ride())

car = Car("Toyota", "Camry", 2023)
print(car.drive(100))
```

```python
class Book:
  def __init__(self, title, author, available):
    self.title = title
    self.author = author
    self.available = available


class Library:
  def __init__(self, books = []):
    self.books = books

  def add_book(self, book):
    self.books.append(book)
    return f"book named `{self.book}` is added to the library"

  def get_available_books(self):
    return [book.title for book in self.books if book.available]

class Member:
  def __init__(self, name, borrowed_books = []):
    self.name = name
    self.borrowed_books = borrowed_books

  def borrow_book(self, book):
    self.borrowed_books.append(book)
    book.available = False
    return f"book named `{book.title}` is borrowed by {self.name}"

  def return_book(self, book):
    self.borrowed_books.remove(book)
    return f"book named `{book.title}` borrowed by {self.name} has now been returned to the library"


book1 = Book("book1", "author1", True)
book2 = Book("book2", "author2", True)
book3 = Book("book3", "author3", True)
book4 = Book("book4", "author4", True)
book5 = Book("book5", "author5", True)
book6 = Book("book6", "author6", True)
library = Library([book1, book2, book3, book4, book5, book6])
member1 = Member("amaan")
member2 = Member("amaan2")

library.get_available_books()
print(member1.borrow_book(book1))
print(member2.borrow_book(book2))
print(member1.return_book(book1))
print(library.get_available_books())
```

```python
class Circle:
    def __init__(self, radius):
        self._radius = radius  # Protected variable

    @property
    def radius(self):
        return self._radius

    @radius.setter
    def radius(self, value):
        if value <= 0:
            raise ValueError("Radius must be positive!")
        self._radius = value

    @property
    def area(self):
        return 3.14 * self._radius ** 2

# Usage
circle = Circle(5)
print(circle.radius)  # Output: 5
circle.radius = 10    # Updates the radius
print(circle.area)    # Output: 314.0
# circle.radius = -5  # Raises ValueError
```

```python
import asyncio

async def task(name, delay):
    print(f"{name}: Starting")
    await asyncio.sleep(delay)  # Simulate I/O
    print(f"{name}: Completed")

async def main():
    await asyncio.gather(
        task("Task 1", 2),
        task("Task 2", 1),
        task("Task 3", 3)
    )

asyncio.run(main())
```

```python
import asyncio

async def long_running_task():
    print("Task started")
    try:
        await asyncio.sleep(10)  # Simulate a long-running task
    except asyncio.CancelledError:
        print("Task was cancelled")
    else:
        print("Task completed")

async def main():
    task = asyncio.create_task(long_running_task())
    await asyncio.sleep(2)  # Let the task run for 2 seconds
    task.cancel()  # Cancel the task
    await task  # Wait for the task to handle the cancellation

asyncio.run(main())
```

```python
import random

async def producer(queue):
  while True:
    number = random.randint(1,10)
    await queue.put(number)
    print(f"producer sent {number}")
    await asyncio.sleep(1)

async def consumer(queue):
  while True:
    number = await queue.get()
    print(f"consumer recieved {number}")

async def main():
  queue = asyncio.Queue()
  await asyncio.gather(producer(queue), consumer(queue))

asyncio.run(main())
```

```python
async def producer(queue):
  while True:
    number = random.randint(1,10)
    await queue.put(number)
    print(f"producer sent {number}")
    await asyncio.sleep(2)

async def task1(queue):
  while True:
    number = await queue.get()
    print(f"worker1 recieved {number}")

async def task2(queue):
  while True:
    number = await queue.get()
    print(f"worker2 recieved {number}")

async def task3(queue):
  while True:
    number = await queue.get()
    print(f"worker3 recieved {number}")

async def main():
  queue = asyncio.Queue()
  await asyncio.gather(producer(queue), task1(queue), task2(queue), task3(queue))

asyncio.run(main())
```

```python
import heapq
from graphviz import Digraph

# Function to visualize the heap
def visualize_heap(heap):
    dot = Digraph()
    for i in range(len(heap)):
        dot.node(str(i), str(heap[i]))  # Add nodes for each element
        left = 2 * i + 1
        right = 2 * i + 2
        if left < len(heap):
            dot.edge(str(i), str(left))  # Connect to left child
        if right < len(heap):
            dot.edge(str(i), str(right))  # Connect to right child
    return dot

# Example heap
heap = [3, 5, 9, 6, 8, 20, 10, 12, 18, 9]
heapq.heapify(heap)

# Visualize the heap
dot = visualize_heap(heap)
dot.render("heap_visualization", format="png", cleanup=True)  # Save as PNG
dot  # Display in Jupyter
```

```python
import timeit

# Compare a list comprehension vs. a loop
setup = "nums = [i for i in range(10000)]"
list_comp = "[x**2 for x in nums]"
loop = """
result = []
for x in nums:
    result.append(x**2)
"""

print("List comprehension:", timeit.timeit(list_comp, setup, number=10000))
print("Loop:", timeit.timeit(loop, setup, number=10000))
```

```python
import asyncio
import aiohttp

async def fetch(url):
    async with aiohttp.ClientSession() as session:
        async with session.get(url) as response:
            return await response.text()

async def main():
    urls = [
        "https://syedamaan.com",
        "https://www.python.org",
        "https://asyncio.org",
    ]

    tasks = [fetch(url) for url in urls]
    results = await asyncio.gather(*tasks)

    for i, content in enumerate(results):
        print(f"Content from URL {i+1}: {content[:500]}...\n")

asyncio.run(main())
```
